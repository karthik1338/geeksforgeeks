#Loading GPT_2 and zero shot model

```python
from transformers import pipeline

# loading gpt-2 model
generator = pipeline("text-generation", model="openai-community/gpt2")

# loading zero_shot_classifer model
zero_shot_classifier = pipeline('zero-shot-classification',model_name='claritylab/zero-shot-explicit-binary-bert')

# segregating the class
candidate_labels = ["other", "python"]
```

#function to generate the code

```python
def python_gpt2(prompt):
  # classification feeding
  classification_results = zero_shot_classifier(prompt, candidate_labels)

  print(f"{classification_results['labels'][0]}: {classification_results['scores'][0]}\n{classification_results['labels'][1]}: {classification_results['scores'][1]}")

  generated_code = generator(prompt, max_length=100, num_return_sequences=1)
  if(classification_results['labels'][0] == 'python'):
      print("-" * 50)
      # gtp-2 feading
      # Corrected: print the generated text and the full response object
      print("Generated Text:")
      print(generated_code[0]['generated_text'])
      print("\nFull Response Object:")
      print(generated_code)

      print("-" * 50)
  else:
      print("-" * 50)
      print("I am not able to provide an answer")
      print("-" * 50)
```

#Prompt 1

```python
prompt = "I am an average man"
python_gpt2(prompt)
```
